```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=funds,
    x="category",
    order=funds["category"].value_counts().index
)

plt.title("Distribution of Mutual Fund Categories")
plt.xlabel("Category")
plt.ylabel("Count")

plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig("../dashboard/fund_category_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=performance,
    x="risk_grade",
    order=performance["risk_grade"].value_counts().index
)

plt.title("Risk Grade Distribution")

plt.tight_layout()

plt.savefig("../dashboard/risk_grade_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=performance,
    x="morningstar_rating"
)

plt.title("Morningstar Ratings")

plt.tight_layout()

plt.savefig("../dashboard/morningstar_rating_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(6,6))

transactions["transaction_type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.title("Transaction Type Distribution")

plt.savefig("../dashboard/transaction_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=transactions,
    x="kyc_status"
)

plt.title("KYC Status")

plt.tight_layout()

plt.savefig("../dashboard/kyc_status.png", dpi=300)

plt.show()
```

```python
fund_house = (
    performance.groupby("fund_house")["aum_crore"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(12,6))

sns.barplot(
    data=fund_house,
    x="fund_house",
    y="aum_crore"
)

plt.xticks(rotation=45)

plt.title("Fund House vs Total AUM")

plt.tight_layout()

plt.savefig("../dashboard/fund_house_aum.png", dpi=300)

plt.show()
```

```python
top_return = performance.sort_values(
    "return_5yr_pct",
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    data=top_return,
    x="return_5yr_pct",
    y="scheme_name"
)

plt.title("Top 10 Schemes by 5-Year Return")

plt.tight_layout()

plt.savefig("../dashboard/top_return.png", dpi=300)

plt.show()
```

```python
top_return = performance.sort_values(
    "return_5yr_pct",
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    data=top_return,
    x="return_5yr_pct",
    y="scheme_name"
)

plt.title("Top 10 Schemes by 5-Year Return")

plt.tight_layout()

plt.savefig("../dashboard/top_return.png", dpi=300)

plt.show()
```

```python
nav["date"] = pd.to_datetime(nav["date"])

monthly_nav = (
    nav.groupby(nav["date"].dt.to_period("M"))["nav"]
    .mean()
    .reset_index()
)

monthly_nav["date"] = monthly_nav["date"].astype(str)

plt.figure(figsize=(15,5))

plt.plot(
    monthly_nav["date"],
    monthly_nav["nav"]
)

plt.xticks(rotation=90)

plt.title("Average Monthly NAV Trend")

plt.tight_layout()

plt.savefig("../dashboard/monthly_nav_trend.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=performance,
    x="std_dev_ann_pct",
    y="return_5yr_pct",
    hue="risk_grade"
)

plt.title("Risk vs Return")

plt.tight_layout()

plt.savefig("../dashboard/risk_return.png", dpi=300)

plt.show()
```

```python
sector = (
    holdings.groupby("sector")["weight_pct"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

sector.plot(kind="bar")

plt.title("Portfolio Sector Allocation")

plt.tight_layout()

plt.savefig("../dashboard/sector_allocation.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(10,8))

corr = performance.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("../dashboard/correlation_heatmap.png", dpi=300)

plt.show()
```
